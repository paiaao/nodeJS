```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.datasets import load_iris

pd.set_option("display.max_columns", None)
```

```python

iris = load_iris(as_frame=True)

df = iris.frame
df["species"] = iris.target_names[df["target"]]
df = df.drop(columns=["target"])
print(df.head())
```

```python
print("Linhas:", df.shape[0])
print("Colunas:", df.shape[1])
print("Nomes:", df.columns.tolist())
```

```python
df.drop(columns=["species"]).hist(figsize=(10,8))
plt.suptitle("Distribuição dos atributos")
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(8,5))
for especie in df["species"].unique():
    dados = df[df["species"] == especie]
    plt.scatter(dados["petal length (cm)"], dados["petal width (cm)"], label=especie)
plt.xlabel("Comprimento da pétala")
plt.ylabel("Largura da pétala")
plt.title("Pétalas por espécie")
plt.legend()
plt.grid(alpha=0.3)
plt.show()
```
